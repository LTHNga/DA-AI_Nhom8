```python
import pandas as pd
```

```python
shopee = pd.read_csv("shopee_data.csv")
books = pd.read_csv("books_crawled.csv")
```

```python
print("Shopee:", shopee.shape)
print("Books:", books.shape)
```

```python
print("Các cột của Shopee:")
print(shopee.columns.tolist())

print("\nCác cột của Books:")
print(books.columns.tolist())
```

```python
shopee_clean = shopee.rename(columns={
    "title": "product_name", #Đổi title sang product_name để thống nhất với books   
    "price_actual": "price",
    "item_rating": "rating",
    "item_category_detail": "category",
    "link_ori": "url"
})

books_clean = books.rename(columns={
    "title": "product_name"
})
```

```python
print("Shopee:")
print(shopee_clean.columns.tolist())

print("\nBooks:")
print(books_clean.columns.tolist())
```

```python
#Tạo category và availability cho shopee_clean và books_clean để thống nhất với nhau
shopee_clean["availability"] = None
books_clean["category"] = None
```

```python
columns = [
    "product_name",
    "price",
    "rating",
    "category",
    "availability",
    "url"
]

shopee_clean = shopee_clean[columns]
books_clean = books_clean[columns]
```

```python
print(shopee_clean.head())
print(books_clean.head())
```

```python
shopee_clean["source"] = "Shopee"
books_clean["source"] = "Books"
```

```python
print(shopee_clean.head())
print(books_clean.head())
```

```python
master = pd.concat(
    [shopee_clean, books_clean],
    ignore_index=True
)
```

```python
print("Kích thước Master Table:", master.shape)
print("\nCác cột:")
print(master.columns.tolist())
```

```python
print(master.info())
```

```python
print(master.isnull().sum())
```

```python
master.to_csv(
    "unified_static_master.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Đã lưu Unified Static Master Table!")
```

```python
import pandas as pd

master = pd.read_csv("unified_static_master.csv")
dynamic = pd.read_csv("dynamic_data.csv")

print("Master:", master.shape)
print("Dynamic:", dynamic.shape)
```

```python
common_products = set(master["product_name"]) & set(dynamic["product_name"])

print("Số product_name trùng nhau:", len(common_products))
```

Mình chọn Left Join vì:

Giữ lại toàn bộ sản phẩm trong master, kể cả những sản phẩm chưa có dữ liệu động. Đây là điểm mình sẽ cần ghi rõ trong report.

```python
merged_data = pd.merge(
    master,
    dynamic,
    on="product_name",
    how="left"
)

print("Kích thước sau merge:", merged_data.shape)
print(merged_data.head())
```

Vì sao tăng lên 31,342?

Vì trong dynamic_data:

300 sản phẩm
mỗi sản phẩm có 30 ngày
→ 9,000 dòng

Khi LEFT JOIN:

20,312 sản phẩm Shopee không có dynamic data → vẫn giữ nguyên 1 dòng/sản phẩm
300 sản phẩm được chọn → mỗi sản phẩm có 30 dòng, thay vì 1 dòng ban đầu.

Nên số dòng là:

20,312 + (300 × 30) = 29,312

Nhưng master còn 300 dòng Books, trong đó có thể có product name trùng/được chọn hoặc duplicate product_name, nên kết quả thực tế lên 31,342. Điều này cho thấy trước khi kết luận merge hoàn toàn đúng, mình nên kiểm tra duplicate key một chút.

```python
print("Master duplicate product_name:",
      master["product_name"].duplicated().sum())

print("Dynamic duplicate product_name:",
      dynamic["product_name"].duplicated().sum())
```

```python
print(
    master.groupby("source")["product_name"]
    .apply(lambda x: x.duplicated().sum())
)
```

```python
selected_products = set(dynamic["product_name"])

master_selected = master[
    master["product_name"].isin(selected_products)
]

print("Số dòng master có sản phẩm được chọn:", len(master_selected))
print(
    "Số product_name khác nhau:",
    master_selected["product_name"].nunique()
)
```

```python
print("Số dòng merged:", len(merged_data))

print("\nSố product_name khác nhau:",
      merged_data["product_name"].nunique())

print("\nSố dòng có dynamic data:",
      merged_data["date"].notna().sum())

print("\nSố dòng không có dynamic data:",
      merged_data["date"].isna().sum())
```

```python
print("Kiểm tra purchase <= add_to_cart:")
print(
    (merged_data["purchase"].dropna() <= 
     merged_data["add_to_cart"].dropna()).all()
)

print("\nKiểm tra add_to_cart <= views:")
print(
    (merged_data["add_to_cart"].dropna() <= 
     merged_data["views"].dropna()).all()
)
```

```python
merged_data.to_csv(
    "merged_data.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Đã lưu merged_data.csv")
print("Kích thước cuối:", merged_data.shape)
```
