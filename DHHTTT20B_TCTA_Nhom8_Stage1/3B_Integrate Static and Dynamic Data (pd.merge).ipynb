```python
import os
import pandas as pd

DATA_DIR = r"D:\L-NMPTDL&AI\Stage 1 + 2\DA-AI_Nhom8\DHHTTT20B_TCTA_Nhom8_Stage1\data"

STATIC_PATH = os.path.join(DATA_DIR, "unified_static_master.csv")
DYNAMIC_PATH = os.path.join(DATA_DIR, "dynamic_data.csv")
```

```python
static_df = pd.read_csv(STATIC_PATH)
dynamic_df = pd.read_csv(DYNAMIC_PATH)

print("Static data:", static_df.shape)
print("Dynamic data:", dynamic_df.shape)

print("\nStatic columns:")
print(static_df.columns.tolist())

print("\nDynamic columns:")
print(dynamic_df.columns.tolist())
```

```python

print("Static product_id duplicates:",
      static_df["product_id"].duplicated().sum())

print("Dynamic product_id missing:",
      dynamic_df["product_id"].isna().sum())

print("Products in static table:",
      static_df["product_id"].nunique())

print("Products referenced by dynamic table:",
      dynamic_df["product_id"].nunique())

# Kiểm tra ID động nào không có trong bảng tĩnh
unmatched_ids = set(dynamic_df["product_id"].dropna()) - set(
    static_df["product_id"].dropna()
)

print("Unmatched product IDs:", len(unmatched_ids))
print("Sample unmatched IDs:", list(unmatched_ids)[:10])
```

```python
merged_df = pd.merge(
    static_df,
    dynamic_df,
    on="product_id",
    how="left",
    suffixes=("_static", "_dynamic")
)

print("Merged data shape:", merged_df.shape)
display(merged_df.head(10))
```

```python
print("Missing values:")
print(merged_df.isna().sum())

print("\nSố lượt tương tác theo nguồn sản phẩm:")
display(
    merged_df.groupby("source")["interaction_id"]
    .count()
    .reset_index(name="interaction_count")
)

print("\nSố sản phẩm có tương tác:")
print(
    merged_df.loc[
        merged_df["interaction_id"].notna(),
        "product_id"
    ].nunique()
)
```

```python
MERGED_PATH = os.path.join(
    DATA_DIR,
    "merged_data.csv"
)

merged_df.to_csv(
    MERGED_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("Đã lưu:", MERGED_PATH)
print("Số dòng:", len(merged_df))
print("Số cột:", len(merged_df.columns))
```
